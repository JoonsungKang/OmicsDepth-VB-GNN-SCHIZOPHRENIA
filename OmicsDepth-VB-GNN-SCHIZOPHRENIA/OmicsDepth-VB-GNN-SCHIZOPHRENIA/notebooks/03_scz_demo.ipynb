# 03 · Schizophrenia multi-omic pipeline on the synthetic demo cohort
Runs the ten steps of `vbgnn/pipeline.py` (main text Sections 4–5) on `data/demo`, a **synthetic** cohort with a
planted mitochondrial–epigenetic feedback loop (`vbgnn/demo.py`). The point is to show every step and to check
recovery of a known loop. Results on synthetic data are not findings about human biology.

To skip training, load the bundled checkpoint: `VBGNN.load("checkpoints/vbgnn_scz_demo.npz")`.

```python
import os, sys
ROOT = os.path.abspath("..") if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
sys.path.insert(0, ROOT)
import numpy as np
import matplotlib.pyplot as plt
```

```python
from vbgnn.pipeline import load_config, run_pipeline
cfg = load_config(os.path.join(ROOT, "configs", "scz_demo.yaml"))
cfg["output_dir"] = "results/scz_demo_notebook"          # keep the committed results untouched
cfg["checkpoint"] = "results/scz_demo_notebook/vbgnn.npz"
run = run_pipeline(cfg)
```

```python
import pandas as pd
run["edges"].head(20)
```

```python
out = run["output_dir"]
pd.read_csv(os.path.join(out, "celltype_flux.csv"))[["group", "targeted_flux"]]
```

```python
pd.read_csv(os.path.join(out, "counterfactual.csv")).pivot(
    index="feature", columns="group", values="fraction_of_gap_recovered").round(2)
```
